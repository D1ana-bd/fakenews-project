```python
import pandas as pd

df_fake = pd.read_csv('../data/raw/fakenewsnet/gossipcop_fake.csv')
df_real = pd.read_csv('../data/raw/fakenewsnet/gossipcop_real.csv')

# Número de tweets por artigo
tweet_counts_fake = df_fake['tweet_ids'].dropna().str.split('\t').str.len()
tweet_counts_real = df_real['tweet_ids'].dropna().str.split('\t').str.len()

print("=== FAKE ===")
print(tweet_counts_fake.describe())
print("\n=== REAL ===")
print(tweet_counts_real.describe())
```
